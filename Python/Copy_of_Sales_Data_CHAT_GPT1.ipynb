```python
# Create a simple sales dataset
sales = [15000, 22000, 18000, 30000, 25000, 27000, 12000]

print("Sales Data:", sales)
```

```python
total_sales = sum(sales)
print("Total Sales:", total_sales)
```

```python
average_sales = sum(sales) / len(sales)
print("Average Sales:", average_sales)
```

```python
highest_sale = max(sales)
lowest_sale = min(sales)

print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

```python
# Count how many days had sales above average
above_average_days = 0

for sale in sales:
    if sale > average_sales:
        above_average_days += 1

print("Number of days with above average sales:", above_average_days)
```

```python
import pandas as pd

# Create a DataFrame
df = pd.DataFrame({
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Sales": sales
})

df
```

```python
df_sorted = df.sort_values(by="Sales", ascending=False)
df_sorted
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["Day"], df["Sales"])
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Weekly Sales Report")
plt.show()
```
