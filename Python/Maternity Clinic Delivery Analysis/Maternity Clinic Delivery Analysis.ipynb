```python
import pandas as pd

# Create maternity clinic dataset
data = {
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Deliveries": [5, 8, 6, 10, 7, 9, 4]
}

df = pd.DataFrame(data)

df
```

```python
total_deliveries = df["Deliveries"].sum()
average_deliveries = df["Deliveries"].mean()
busiest_day = df.loc[df["Deliveries"].idxmax()]
least_busy_day = df.loc[df["Deliveries"].idxmin()]

print("Total Deliveries:", total_deliveries)
print("Average Deliveries per Day:", average_deliveries)
print("\nBusiest Day:")
print(busiest_day)

print("\nLeast Busy Day:")
print(least_busy_day)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(df["Day"], df["Deliveries"])
plt.xlabel("Day")
plt.ylabel("Number of Deliveries")
plt.title("Weekly Maternity Deliveries")
plt.show()
```

```python
#30-Day Maternity Ward Dataset
```

```python
import pandas as pd
import numpy as np

# Set random seed for consistent results
np.random.seed(42)

# Create 30 days of deliveries (between 4 and 12 per day)
deliveries = np.random.randint(4, 13, 30)

# Create DataFrame
df_30 = pd.DataFrame({
    "Day": range(1, 31),
    "Deliveries": deliveries
})

df_30.head()
```

```python
total_deliveries = df_30["Deliveries"].sum()
average_deliveries = df_30["Deliveries"].mean()
max_day = df_30.loc[df_30["Deliveries"].idxmax()]
min_day = df_30.loc[df_30["Deliveries"].idxmin()]

print("Total Deliveries in 30 Days:", total_deliveries)
print("Average Deliveries per Day:", round(average_deliveries, 2))
print("\nHighest Delivery Day:")
print(max_day)

print("\nLowest Delivery Day:")
print(min_day)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df_30["Day"], df_30["Deliveries"])
plt.xlabel("Day of Month")
plt.ylabel("Number of Deliveries")
plt.title("30-Day Maternity Ward Deliveries")
plt.show()
```

```python
# Count how many days had 10 or more deliveries
high_workload_days = df_30[df_30["Deliveries"] >= 10].count()["Deliveries"]

print("Number of High Workload Days (>=10 deliveries):", high_workload_days)
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Total deliveries per day (same as before)
deliveries = np.random.randint(4, 13, 30)

# Create C-section percentage (between 20% and 40%)
c_section_percent = np.random.uniform(0.2, 0.4, 30)

# Calculate C-sections and Normal deliveries
c_sections = (deliveries * c_section_percent).round().astype(int)
normal_deliveries = deliveries - c_sections

df_comp = pd.DataFrame({
    "Day": range(1, 31),
    "Total Deliveries": deliveries,
    "C-Sections": c_sections,
    "Normal Deliveries": normal_deliveries
})

df_comp.head()
```

```python
total_deliveries = df_comp["Total Deliveries"].sum()
total_c_sections = df_comp["C-Sections"].sum()
total_normal = df_comp["Normal Deliveries"].sum()

c_section_rate = (total_c_sections / total_deliveries) * 100

print("Total Deliveries:", total_deliveries)
print("Total C-Sections:", total_c_sections)
print("Total Normal Deliveries:", total_normal)
print("C-Section Rate (%):", round(c_section_rate, 2))
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(["Normal Deliveries", "C-Sections"], 
        [total_normal, total_c_sections])
plt.xlabel("Delivery Type")
plt.ylabel("Total Cases")
plt.title("Monthly Delivery Type Distribution")
plt.show()
```

```python
import numpy as np

np.random.seed(42)

# Simulate complication rates
pph_rate = np.random.uniform(0.03, 0.08, 30)  # 3% - 8%
preeclampsia_rate = np.random.uniform(0.05, 0.12, 30)  # 5% - 12%

# Calculate complication counts
pph_cases = (df_comp["Total Deliveries"] * pph_rate).round().astype(int)
preeclampsia_cases = (df_comp["Total Deliveries"] * preeclampsia_rate).round().astype(int)

# Add to dataframe
df_comp["PPH Cases"] = pph_cases
df_comp["Pre-eclampsia Cases"] = preeclampsia_cases

df_comp.head()
```

```python
total_pph = df_comp["PPH Cases"].sum()
total_preeclampsia = df_comp["Pre-eclampsia Cases"].sum()

pph_rate_month = (total_pph / total_deliveries) * 100
preeclampsia_rate_month = (total_preeclampsia / total_deliveries) * 100

print("Total PPH Cases:", total_pph)
print("Total Pre-eclampsia Cases:", total_preeclampsia)
print("Monthly PPH Rate (%):", round(pph_rate_month, 2))
print("Monthly Pre-eclampsia Rate (%):", round(preeclampsia_rate_month, 2))
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(["PPH", "Pre-eclampsia"], 
        [total_pph, total_preeclampsia])
plt.xlabel("Complication Type")
plt.ylabel("Total Cases")
plt.title("Monthly Obstetric Complications")
plt.show()
```
